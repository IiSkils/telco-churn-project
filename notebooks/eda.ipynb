```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. قراءة البيانات
df = pd.read_csv('../data/telco_churn.csv')

# 2. نظرة عامة
print(df.info())
display(df.head())

# 3. فحص توازن الفئات (Churn)
plt.figure(figsize=(6, 4))
sns.countplot(data=df, x='Churn', palette='Set2')
plt.title('Target Variable Distribution (Churn)')
plt.show()

# 4. حساب نسبة عدم التوازن
churn_ratio = df['Churn'].value_counts(normalize=True) * 100
print(f"Churn Ratio:\n{churn_ratio}")
```
